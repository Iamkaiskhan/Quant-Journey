```python
# my code explanation with comments

import numpy as np # Note: NumPy is imported but not utilized in this specific script

# 'arr' is the 1000-digit dataset you are analyzing
arr = '7316717653133062491922511967442657474235534919493496983520312774506326239578318016984801869478851843858615607891129494954595017379583319528532088055111254069874715852386305071569329096329522744304355766896648950445244523161731856403098711121722383113622298934233803081353362766142828064444866452387493035890729629049156044077239071381051585930796086670172427121883998797908792274921901699720888093776657273330010533678812202354218097512545405947522435258490771167055601360483958644670632441572215539753697817977846174064955149290862569321978468622482839722413756570560574902614079729686524145351004748216637048440319989000889524345065854122758866688116427171479924442928230863465674813919123162824586178664583591245665294765456828489128831426076900422421902267105562632111110937054421750694165896040807198403850962455444362981230987879927244284909188845801561660979191338754992005240636899125607176060588611646710940507754100225698315520005593572972571636269561882670428252483600823257530420752963450'
# 'arr1' is a shorter test string (currently unused in the main logic below)
arr1 = '12345678000545789064465'

# Sets the size of the "sliding window" you want to analyze (e.g., 13)
number = int(input('Enter the number:'))

max_sum = 0
number1 = 0  # This will store the actual digit sequence that produces the highest sum

# OUTER LOOP: Moves the starting position of your window one digit at a time.
# 'len(arr) - number + 1' prevents an IndexError by ensuring the loop stops 
# when there are exactly 'number' digits left at the end of the string.
for i in range(len(arr) - number + 1):
    k = ''           # Clears 'k' to hold the new chunk of digits for this iteration
    status = True    # A flag to track if the current sequence is valid (contains no zeroes)
    
    # INNER LOOP: Builds the sequence character by character
    for j in range(i, i + number):
        k += arr[j]  # Extracts the character at index 'j' and appends it to string 'k'
        
        # Optimization: If a '0' is detected, flag it as False and break the inner loop early.
        # This saves processing time since any sequence with a zero has a product of zero.
        if arr[j] == '0':
            status = False
            break
            
    t = 0
    
    # If the sequence was valid (no zeroes were found):
    if status:
        # Loop through the extracted string 'k' digit by digit
        for x in k:
            t += int(x)  # Convert the string character back to an integer and add it to sum 't'

    # Check if the current valid sequence has set a new sum record:
    if status:        
        if t > max_sum:
            max_sum = t         # Replace the old max_sum with the new highest sum
            number1 = int(k)    # Save the sequence itself (converted to an integer) so we can multiply it later

product = 1

# FINAL LOOP: Takes the 'winning' integer (number1) that had the highest sum, 
# converts it back to a string so it can be iterated through, and multiplies each digit.
for i in str(number1):
    product *= int(i)

print(max_sum)  # Outputs the highest sum found in the 1000-digit string
print(product)  # Outputs the product of the sequence that yielded that highest sum
print(number1)  # Outputs the winning sequence itself
```
