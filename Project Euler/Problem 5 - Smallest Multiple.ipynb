```python
# the actual way

from math import log 
import numpy as np

k = int( input( 'Enter the number' ) ) 

# first talk about capturing the prime numbers

def prime_list( k ) :
    is_prime = [ True ] * ( k+1 ) # (k+1 ) so that we can treat the indexes as the number

    is_prime[0] = is_prime[1] = False  # manually set them false
    
    for p in range ( 2 , int( k ** 0.5 ) + 1 ) :

        if is_prime[ p ] :   # it means that number is a prime
            
            for multiple in range( p*p , k+1 , p ) :
                is_prime[multiple] = False # so it will set the multiple of prime number as composite number as obvious


    prime = [ i for i in range( 0 , k+1 ) if is_prime[i] ]
    print ( prime )



    # now our next goal is to found the smallest multiple with the help of this prime factors
    prime_pow = [] 
    for i in range(len(prime)) :
        status = True 
        a = 0
        
        while status :
            if pow( prime[i] , a ) <= k :
                a+=1

            else :
                prime_pow.append( a - 1 ) 
                status = False

    print( prime_pow )

    prime1 = np.array ( prime )
    prime_pow1 = np.array( prime_pow )

    arrayy = pow( prime1 , prime_pow1 )
    product = 1
    for i in arrayy :
        product *= i

    print( product )
# now our next goal is to found the smallest multiple with the help of this prime factors
prime_list( k ) 
```
