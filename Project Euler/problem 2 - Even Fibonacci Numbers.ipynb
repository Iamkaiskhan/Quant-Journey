# Problem 2 - Even Fibonacci Numbers

```python
def fibonacci(x):
    if x == 1 :
        return 1
    elif x == 2 :
        return 2
    else:
        return fibonacci(x-1) + fibonacci(x-2)
x = 1
list1 = []
while True:

    if fibonacci(x) < 4000001:
        list1.append( fibonacci(x) )
        x+=1
        fibonacci(x)

    else:
        break
sum = 0
for i in range(len(list1)):
    if list1[i] % 2 == 0:
        sum += list1[i]

print( sum )
        
```
