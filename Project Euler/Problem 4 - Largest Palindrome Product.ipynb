```python
import numpy as np

a = 100
var = np.zeros( ( 1 , 3 ) )

for i in range( a , 1000 ) :
    for j in range ( a+1 , 1000 ) :
        k = i * j
        str1 = str(k)
        str2 = ''
        for x in range(len(str1)) :
            str2 = str2 + str1[-x-1]

        if str1 == str2 :
            var = np.vstack( ( var , [[ int(str1) , i , j ]] ))
            

    a+=1

print( var[1228] )
```
