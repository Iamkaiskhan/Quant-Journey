```python
# lets learn about dataframe
import pandas as pd

s1 = [ 1 , 2 , 3 , 4 , 5 ]

var1 = pd.DataFrame( s1 )

print( var1 ) 
print()
print( type(var1) )
```

```python
# lets pass it through the dictionary 
# while passing the 'value' to the 'key' the 'value' length must be same of the 'key'
s2 = {
    'a' : [ 1 , 2 , 3 , 4 , 5 ] , 
    's' : [ 11, 12 , 13 , 14 , 15 ]
}

var2 = pd.DataFrame( s2 )
print( var2 )
```

This is the error if we have unequal value length in dictionary
<br>


```python
# if we want the output for the particular column 

s3 = {
    'a' : [ 1 , 2 , 3 , 4 , 5 ] , 
    's' : [ 11, 12 , 13 , 14 , 15 ] , 
    1 : [ 6 , 3 , 4 , 7 , 3 ]
}

var3 = pd.DataFrame( s3 , columns = [ 'a' , 1 ] )  # we will give input in columns as list of the column that we want to extract
print( var3 )
```

```python
# now if we want to change the index of the row as well 

s4 = {
    'a' : [ 1 , 2 , 3 , 4 , 5 ] , 
    's' : [ 11, 12 , 13 , 14 , 15 ] , 
    1 : [ 6 , 3 , 4 , 7 , 3 ]
}

var4 = pd.DataFrame( s4 , columns = [ 'a' , 1 ] , index = [ 'a' , 'b' , 'c' , 'd' , 'e' ] )
print( var4 )
```

```python
# now to get the particular element 

s5 = {
    'a' : [ 1 , 2 , 3 , 4 , 5 ] , 
    's' : [ 11, 12 , 13 , 14 , 15 ] , 
    1 : [ 6 , 3 , 4 , 7 , 3 ]
}

var5 = pd.DataFrame( s4 , columns = [ 'a' , 1 ] , index = [ 'a' , 'b' , 'c' , 'd' , 'e' ] )
print( var5 )
print()

print( var5['a']['c'] )  # like this 'var5[][]' we can extract the particular element

```

```python
# now we will make dataframe with the help of the list

list6 = [[1,2,3,4,5],[11,12,13,14,15]]

var6 = pd.DataFrame( list6 )

print( var6 )
```

```python
# we can also create DataFrame with combination of dictionary and series data structure

dic7 = {
    'a' : pd.Series([ 1,2,3 ]) , 'b' : pd.Series([4,5,6])
}

var7 = pd.DataFrame( dic7 )

print( var7 )
```
