```python
import pandas as pd

x = [ 1,2,3,4,5 ] 

a = pd.Series(x)  # it will store the data in 1 D which means as series

print( a )

print()

print( type(a) )
print()

# to call any element by index
print( a[2] )
```

```python
# now we will learn about indexing 

b = [ 2 , 3 , 4 ]

# In the below variable , we have set index as per our need , datatype as per our need and name of the data structure as per our need
y = pd.Series( b , index = [ 'a','b','c' ] , dtype = 'float' , name = 'sample' )   # we can also give index as per our requirement

print ( y ) 
 
```

```python
# Lets pass dictionary in the pandas
dic = {
    'name' : [ 'python' , ' java ' , ' c++ ' ] , 'rank' : [ 1 , 3 , 2 ] , 'intrest' : [ 'High' , 'Medium' , 'Low' ]
}

var = pd.Series( dic )
print( var )
```

```python
# making series with single element

var1 = pd.Series(12)
print( var1 )
print()

# now we will learn to place single element at multiple places

var2 = pd.Series( 20 , index = [ 1,2,3,4,6 ] )  # it will place the ' 20 ' at the given indexes
print( var2 )
print()

var3 = pd.Series( 13 , index = [ 1, 2, 3 ] )
print( var3 )
print() 

# now we will add this this dataframes , in case of numpy this would have been broadcasting error
# in case of pandas it will show NaN as output in case if the index values are not found , hence it works with missing data as well

print( var2 + var3 )
print()
```
