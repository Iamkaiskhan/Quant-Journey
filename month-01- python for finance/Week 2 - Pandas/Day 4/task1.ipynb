```python
import pandas as pd 

csv = pd.read_csv( 'C:\\Users\HP\\Documents\\GitHub\\Quant-Journey\\month-01- python for finance\\Week 2 - Pandas\\Day 3\\Ambition Box.csv' , nrows = 20 , usecols = [ 0 , 1 , 2 , 3 , 4 ] )
csv = csv.rename( columns = { 'Unnamed: 0' : 'Series' } )

csv = csv.set_index('Series') #  it is used to set the particular column to behave as index

print( csv ) 
```

```python
# lets learn to know how many rows are there in the data structure ( the index inital and final )

a = csv.index
print( a ) 

# to know the names of the columns

b = csv.columns
print( b )

# to know the min , max ,  std dev , mean , etc
c = csv.describe()
print( c )
```

```python
# to get the top n datas , we can use the head(n) function

d = csv.head(6)  # it will print the top 6 data
print( d )                          
print()

# to get the datas of last n rows , we can use the tail(n)

e = csv.tail( 3 )
print(e)
print()
```

```python
# lets learn how to get the particular data
## to get the detail of the partticular column
a1 = csv[[ 'name' , 'rating' ]][2:5]
print( a1 )
print()


## to get the data of particular rows
b1 = csv[ 2:10 ]
print( b1 )
```

```python
# we can also convert the pandas to numpy 

a2 = csv.to_numpy()
print( a2 )
```

```python
# to sort the dataframe in descending order across the row

a3 = csv.sort_index( axis = 0 , ascending = False ) # axis 0 means across the rows and axis 1 means across the column
print( a3 ) 
print()
```

```python
# now we can also change the particlar cell in the DataFrame

csv['name'][0] = 'Kais' # it will change the cell data
print( csv ) 
print() 

csv.to_csv('Test.csv')


# we can also use the loc() funtion as well to change the cell inputt

csv.loc[0,'name'] = 'Qais'
print( csv ) 
print()
```

```python
# now we can also use loc() to get the particular data

a4 = csv.loc[ [2,5] , [ 'name' , 'rating' ]]   # it will print the data of that 'name' and 'rating' column of row 2 and 5 only
print( a4 ) 
```
