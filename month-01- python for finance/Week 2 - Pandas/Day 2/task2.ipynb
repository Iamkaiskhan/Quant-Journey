```python
# Lets learn about insert and delete function

import pandas as pd

var1 = pd.DataFrame( { 'A' : [ 1,2,3 ] , 'B' : [ 4,5,6 ] } )
print( var1 )
print()

var1.insert( 1 , 'Python' , var1['A'] )   # insert( column index , column name , element values )  , the element length must be same
print( var1 )
```
