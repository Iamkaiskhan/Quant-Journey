# Arithmatic Operations

```python
import pandas as pd

s1 = {
    'A': [ 1,2,3,4,5 ] , 'B' : [ 6,7,8,9,10 ] 
}

var1 = pd.DataFrame( s1 )

print( var1 )
print()

# now lets make a column 'C' in which it will have summation of column A and B 

var1['C'] = var1['A'] + var1['B']

print( var1 )
print()

# multiplication

var1['C'] = var1['A'] * var1['B']

print( var1 )
print()

# division

var1['C'] = var1['A'] / var1['B']

print( var1 )
print()
```

```python
# Lets learn about logical operation

s2 = {
    'A' : [ 33, 24 ,24 , 21 , 15 , 76 ] ,
    'B' : [ 12, 44, 65, 23 , 29, 88 ]
}

var2 = pd.DataFrame( s2 )
print( var2 )
print()


# lets test if the value of A column is >20

var2['A>20'] = ( var2['A'] > 20 )
print( var2 )
```
