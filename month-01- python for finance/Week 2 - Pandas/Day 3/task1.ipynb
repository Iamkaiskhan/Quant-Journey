```python
import pandas as pd


# lets learn about inserting columns in DataFrame

var = pd.DataFrame( { 'A' : [ 1 , 2 , 3 , 4 , 5 ] , 'B' : [ 6 , 7 , 8 , 9 , 10 ] } )

print( var )
print()

var.insert( 1 , 'Z' , [ 11,12,13,14,15] )
print( var )
print()

# lets learn about creating column in the last index

var['ZZ'] = var['A'][ : 3 ]
print(var)
print()
```

```python
# now lets learn to delete the colunn , for that we will use the 'pop' function 

var1 = var.pop('ZZ')
print(var1)
print()

print(var)
print()

# method to insert the pop ZZ again
## method 1 
var['ZZ'] = var1
print(var)
print()


del var['ZZ']

## method 2 
var.insert( 0 , 'ZZ' , var1 )
print( var )
```
