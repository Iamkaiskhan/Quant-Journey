```python
# Write csv

import pandas as pd

dis = {
    'a': [ 1,2,3,4,5 ] , 
    'b': [ 11,12,13,14,15]
}

var = pd.DataFrame( dis )

print( var )
print()



var.to_csv( 'test.csv' )   # to_csv function is used to convert the file to csv
```

```python
# since the above file has index also in it , so we will remove that first
import numpy as np 


dis1 = {
    'a' : np.random.randint( 0 , 20 , 10 ) ,
    'b' : np.random.randint( 20 , 40 , 10 )
}

var1 = pd.DataFrame( dis1 )

print( var1 )
print() 


var1.to_csv( 'test.csv' , index = False ) 
```



<br>

see the index is removed now 

```python
# similarly we can also change the name of the columns when converting it to csv

dis2 = {
    1 : np.random.randint( 0 , 20 , 10 ) , 
    2 : np.random.randint( 20 , 40 , 10 ) ,
    3 : np.random.randint( 40 , 60 , 10 )
}

var2 = pd.DataFrame( dis2 )
print( var2 ) 
print() 

var2.to_csv( 'test.csv' , index = False , header = [ ' Column A ' , ' Column B ' , ' Column C ' ] )
```
