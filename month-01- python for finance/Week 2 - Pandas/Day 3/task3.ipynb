```python
# we will learn to read the csv files

import numpy as np 
import pandas as pd

csv1 = pd.read_csv( 'Ambition Box.csv' )

print( csv1 )
```

```python
# to get a particular rows
csv1 = pd.read_csv( 'Ambition Box.csv' , nrows = 10 , usecols = [ 'name' , 'rating' , 'company_type' ] ) 
# nrows function allows us to extract the rows from intital index 0 to the index that we want 
# usecols allows us to filter the DataFrame with the columns that we want , usecols = [ 0 , 3 ] it also works it will filter the column of index 0 and index 3

 

print( csv1 ) 
```

```python
csv2 = pd.read_csv( 'Ambition Box.csv' , nrows = 10 , index_col = "name" , usecols = [ 'name' , 'rating' , 'company_type' ] )
# index_col function allows us to convert that particluar columns into heading

print( csv2 ) 
print()
print( csv2['rating'] )
```

```python
csv3 = pd.read_csv( 'Ambition Box.csv' , nrows = 20 , usecols = [ 0 , 1 , 2 , 3 , 4 ] , dtype = { 'Unnamed: 0' : 'float' } )

csv3 = csv3.rename( columns = { 'Unnamed: 0':'Series' } ) # we can use this to rename the column
 
print( csv3 )
```
