```python
import pandas as pd

csv = pd.read_csv( r"C:\Users\HP\Downloads\student-dataset.csv" , nrows = 60 )

csv['id'] = pd.to_numeric( csv['id'] , errors = 'coerce' )

csv = csv.set_index( 'id' )

print( csv ) 
print() 

# now we will drop the datas which have the NaN values in it 

csv = csv.dropna()
print( csv )
print() 

# to convert the dtype of an index

csv.index = csv.index.astype( int ) 
print( csv )
print() 
```
