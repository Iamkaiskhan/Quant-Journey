# Maps

```python
# instead of writing for loop we use .map() function

a_points_mean = a.points.mean()

a.points.map( lambda p : p - (a_points_mean) )
```

```python
# we can simply take the difference as well
# this is more faster 
a.points - a.points.mean()
```

```python
# .map() is used to apply on a single column only , therefore it requires to give the column name
# map either take dictionary or lambda function 
a.country.map( { 'France' : 'Fra' , 'Italy':'Ity' , 'New Zealand' : 'NZ' } )  # see here we worked only in country column
```

```python
# now we will learn to use .apply , apply is when we have to deal with multiple columns 
# for this first make a fuction of ur required rules 

def score(row) :  # row is the dataframe
    if row.points >= 80 and row.price < 50 :
        return 10

    else :
        return 5


a.apply( score , axis = 'columns' )  # see here we directly applied the .apply on 'a' dataframe 

# axis = 1 or axis = 'columns' means it will cross check each and every row one by one
# axis = 0 or axis = 'index' means it will cross check each colunmn one by one 
```

```python
# we can also use the .apply for a particular column 
k = a.apply( lambda row : row.points/row.price , axis = 1 )

# below i am just checking the NaN records for k = NaN
k.loc[ k.isnull() ]
```

```python
# below is the example how to use apply for index i.e axis as 0

# 1. Define a rule that works on a WHOLE column
def get_spread(column):
    # This function expects a massive vertical list of numbers
    return column.max() - column.min()

# 2. Select only the numerical columns from your dataset
numerical_data = a[['points', 'price']]

# 3. Apply the rule going vertically (axis='index')
numerical_data.apply(get_spread, axis='index')

```
