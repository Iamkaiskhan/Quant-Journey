```python
import pandas as pd

reviews = pd.read_csv(r"C:\Users\HP\Documents\GitHub\Quant-Journey\month-01- python for finance\Week 2.0 - Pandas\winemag-data-130k-v2\winemag-data-130k-v2.csv")
```

```python
desc_agg = reviews.groupby(['country','province']).description.agg([len])
```

```python
# now we will sort this 
regular_desc = desc_agg.reset_index()
regular_desc
```

```python
# sorting it 
regular_desc.sort_values(by = 'len')   # here 'len' is the name of the column
```

```python
# by default it will sort this in ascending now to sort it in descending

descending_format = regular_desc.sort_values( by = 'len' , ascending = False )

descending_format
```

```python
# now to sort it by index use , sort_index()
descending_format.sort_index()
```

```python
# we can sort it by multiple columns as well 

description_multi = descending_format.sort_values( by = ['country','province'] )
description_multi
```
