```python
# now we will learn about the combining function

import pandas as pd 

canadian_yt = pd.read_csv(r"C:\Users\HP\Documents\GitHub\Quant-Journey\month-01- python for finance\Week 2.0 - Pandas\CA_youtube_trending_data.csv", index_col = 0 , nrows = 500)
british_yt = pd.read_csv(r"C:\Users\HP\Documents\GitHub\Quant-Journey\month-01- python for finance\Week 2.0 - Pandas\BR_youtube_trending_data.csv" , index_col = 0 , nrows = 500 )
```

```python
# we use .concat() when we have the column name in the csv's as same i.e. it just add up the rows
pd.concat( [ canadian_yt , british_yt ] )
```

```python
# now we will use the join()

left = canadian_yt.set_index(['title' , 'trending_date'])
left
```

```python
right = british_yt.set_index(['title','trending_date'])
right
```

```python
# now we will use the join() function which will join based on the rows index 

# I have created a variable 'left' which have the index 'title' and 'trending_date'
# I have created a variable 'right' which have the index 'title' and 'trending_date'

left.join( right , lsuffix = '_CAN' , rsuffix = '_UK' ) 
```
