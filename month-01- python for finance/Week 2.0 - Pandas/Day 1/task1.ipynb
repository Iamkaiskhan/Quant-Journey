```python
import pandas as pd

a = pd.DataFrame({'Yes':[ 12 , 31 ] , 'No' : [55 , 13]})

a
```

```python
# we can also use the strings in it 

b = pd.DataFrame({ 'Bob':['Hey i am bob','I am a trader'] , 'Alice':['Hey i am alice', ' I am a analyst'] } )

b
```

```python
# we can also set the index as per our need

c = pd.DataFrame({ 'Bob':[ 'I am bob' , ' i am a trader '] , 
                   'Alice' : [ 'I am alice' , ' i am a analyst'] }, 
                   index = [ ' Introduction ' ,'Profession'] )

c
```

```python
# now we will learn about series 
# series can be created by the use of list and it dosent have column name but it has the overall name

d = pd.Series( [ 30 , 40 , 50 ] )
d
```

```python
# we can also change the name of the index and we can also give the overall name to the series 

e = pd.Series( [ 30 , 45 , 60 ] , index = [ ' 2024 sale ' , ' 2025 sale ' , ' 2026 sale ' ] , name = 'product' )

e
```
