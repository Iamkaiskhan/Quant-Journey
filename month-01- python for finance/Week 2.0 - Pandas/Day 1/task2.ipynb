# iloc Function

```python
# now we will learn about iloc which means index based allocation based on numerical positon 

a.iloc[0] # it will print the data of row 0 in series data structure
```

```python
# its first row and second column i.e iloc[row][column]
a.iloc[ : , 0 ]
```

```python
# we can also use it as :
a.iloc[ 1 : 3 , 0 ]
```

# loc function

```python
# loc is the label based selection 

a.head()
```

```python
list2 = [ 'country' , 'title' , 'winery' , 'points' ]
a.loc[ 0 , list2 ] # simply taking output of single row , here it will give output as series datastructure
```

```python
a.loc[ : , list2 ]
```

```python
# there is very huge difference in loc and iloc 
# since the index can be in string as well and in that case we cant use the iloc funtion we have to use loc 

# also loc is inclusive that in 1:10 it will print all row with index from 1 to 10 , iloc is exculisve i.e it will print from index 1 to 9
# and this is because the index can be string so it has to include the last string row index
b = a.iloc[0]  # its series
print( b )

b.loc[ 'country':'title' ]   # this is how it is used
```

```python
# to get a particular value then first convert the required column to series then use the iloc or loc
a.description.iloc[0]

a.description.loc[0]

a.description[0]
```

# Conditional selection

```python
# like if want to filter a column whose value is equals something
a.country == 'Italy'
```

```python
# we can also use the  loc for this task , it will directly print the datas whose value comes True
a.loc[a.country == 'Italy' ] # we cant write the index here it will show error
```

```python
# now we will also like to see the wine which are more than average , let say wines with points >= 80

a.loc[ (a.country == 'Italy') & (a.points >= 80) ]   # we have to use & not and
```

```python
# we can also use the | it means either of one is true
a.loc[ ( a.country == 'Italy') | ( a.points >=80 ) ]
```

```python
# we can also use .isin function instead of |
list3 = [ 'France' , 'Italy' , 'US' ]   
a.loc[ a.country.isin( list3 ) ]

# if we had used | 
# a.loc[ ( a.country = 'France' ) | ( a.country == 'Italy') | ( a.country == 'US' ) ]

```

```python
# we can use either .isnull() or .notnull() to either see or ignore the null related values

# using .isnull()

a.loc[ a.price.isnull() ]
```

```python
# using .notnull()
a.loc[ a.price.notnull() ]
```
