```python
# lets learn about iteration in numpy

import numpy as np

arr = np.array ( np.arange(10) )
print(arr)
print(len(arr))
for i in arr:
    print(i)
```

```python
# iteration in 2 d

arr1 = np.array( [ np.random.randint(0,100,10) , np.random.randint(0,100,10) ])
print(arr1)
print()

# method of iteration for printing row completely
for i in arr1:    # this will acutally print row
    print(i)

print()
# method of iteration for printing the elements inside a row nd column

for x in arr1:
    for y in x:
        print(y)
```

```python
# lets learn it for 3 d array

arr2 = np.array ( [[ np.random.randint(0,100,8) , np.random.randint(0,100,8) ] , [ np.random.randint(0,100,8) , np.random.randint(0,100,8) ]] )

print(arr2)

for x in arr2:   # it will select the depth , in this case it will run two times due to depth is 2 in arr2
    for y in x:  # it will then select the rows , here total there are 2 rows in each depth 
        for z in y: # it will select the columns , here  
            print(z)
            
```

```python
arr3 = np.array ( [[ np.random.randint(0,100,8) , np.random.randint(0,100,8) ] , [ np.random.randint(0,100,8) , np.random.randint(0,100,8) ]] )

# we can also use the built in functions for printing elements
print(arr3)
print()

for i in np.nditer(arr3):
    print(i)


# if u also want to know the index then
for j,k in np.ndenumerate(arr3):
    print(j,k)
    
```
