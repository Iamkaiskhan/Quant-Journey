```python
# lets learn about copy and view

import numpy as np

var = np.array( np.random.randint( 1,10,5 ) )
print( var ) 
print()

# to copy
x = var.copy()
print( x )


# to copy 
y = var.view()
print()
print( y )
```

```python
# lets see the difference in view() and copy()
var1 = np.array( np.random.randint(0,10,6) )

x = var1.copy()
var1[2] = 30

print( 'var1',var1)
print()
print('x' ,x)
print()

y = var1.view()
print('y',y)
```
