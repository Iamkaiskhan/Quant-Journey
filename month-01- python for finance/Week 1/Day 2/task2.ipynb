```python
# lets learn about rand() function in random module

import numpy as np

val = np.random.rand(6)    # 6 is the total elements in the array and each element will have value between 0 to 1

print(val)

val1 = np.random.rand(3,5)     # it will give array with 3 rows and 5 columns and it will be 2-D array
print(val1)


# lets learn about randn() function , it gives values which are close to zero

val2 = np.random.randn(3,5) # the number can be postive or negative
print(val2)
```

```python
# lets learn about ranf() function ,its range is [0.0,1.0)

val3 = np.random.ranf(5)
print(val3)
```

```python
# lets learn about randint() function so randint(min,max,total elements)   here min is inclusive but max is exclusive

val4 = np.random.randint(5,60,50)
print(val4)
```

```python
import numpy as np

var = np.array([1,2,3,4])
print('Data Type:',var.dtype)        # we have to use dtype() function for this
```

```python
var1 = np.array([3.4,4,6.3])       # it will return data type as float
print("Data Type:",var1.dtype)
```

```python
var2 = np.array(['S','K','D'])
print("Data Type:",var2.dtype)
```

```python
# lets do the combination of string and integers
var3 = np.array(['s','k','a','e',3,45,6,3,4])
print('Data Type:',var3.dtype)
```

```python
# lets learn to convert data types

x = np.array([3,4,5,3,0] , dtype = np.int8)
print('Data Type:',x.dtype)
```

```python
 # to convert let say from int to float

y = np.array([2,3,4,5,6],dtype='f')
print('Data Type:',y.dtype)
```

```python
z = np.array([3,4,5,2,5,6])
print('Data Type:',z.dtype)

new_z = np.float64(z)    # we can pass the array through the required data type to change the data type
print(z,'\n',new_z.dtype,sep='')

new = np.int_(new_z)
print(new,'\n',new.dtype,sep='')
```

```python
# we can also use the astype() function for the data type conversion
a = np.array([5,3,6,3,677,3,56])
a_new = a.astype(float)

print(a,'\n',a_new,sep='')
```
