```python
# lets learn about arithmatic functions
import numpy as np
from math import pi
array = np.array([5,43,5,3,5,65,9,pi])

print(np.min(array))   #  min() - it will return the value of element with minimum value
print(np.max(array))  #   max() - it will return the value of element with maximum value
print(np.argmin(array)) # argmin() - it will tell the position of min element with index intial 0
print(np.argmax(array))   # argmax() -  it will tell the position of max element with index intial 0
print(np.cumsum(array))   #  cumsum() - it will return us the cumulative sum of the elements
print(np.sqrt(array))   #   sqrt() - it will return the value of square root of the array's elements
print(np.sin(array))   #  sin() - it is a sin function with value in radian
print(np.cos(array))  #  cos() - it is a cos function with value in radian 
```

```python
# lets learn about the 2 - D arrays with this function
# axis = 0 means across columns
# axis = 1 means across rows

array2d = np.array([[4,5,623,6,32,56,9] , [3,4,56,6,4,76,4]])
print( np.min( array2d , axis = 0 ))   # it will check the matrix column wise
print( np.max( array2d , axis = 1 ))   # it will check the matrix across the row 
```
