# lets learn about arithmetic operations




```python
import numpy as np

array1 = np.array([2,3,5,3,5,3,9])

array_add1 = array1 + 5
print(array_add1)

array2 = np.array([4,5,6,3,6,34,8])   # the number of elements in all the arrays should be same 

array_add2 = array1 + array2   # similarly we can perform for subtraction , multiplication , division , modulo , power , reciprocal
print(array_add2)
```

```python
# lets use function for this arithmatic operation
array1 = np.array([2,3,5,3,5,3,9])
array2 = np.array([4,5,6,3,6,34,8])

print(np.add(array1,array2))
print(np.subtract(array1,array2))
print(np.multiply(array1,array2))
print(np.divide(array1,array2))
print(np.power(array1,array2))
print(np.mod(array1,array2))
print(np.reciprocal(np.float64(array1)))   # we have to first convert the data type of int to float
```

```python
# we have to follow the similar manner for 2-D arithmatic operations

array1 = np.array([[2,3,5,3,5,3,9],[4,67,4,56,4,3,9]])
array2 = np.array([[4,5,6,3,6,34,8],[3,4,234,4,2,34,0]])

print(array1 + array2) # similarly for other arithmatic operations and we can also use arithmatic functions


```
