```python
import numpy as np

var = np.matrix( [ np.random.randint(0,20,5) , np.random.randint(20,40,5) ] )

print( var )
print( type(var) )
```

```python
var1 = np.matrix( [ np.random.randint(0,20,5) , np.random.randint(20,40,5) ] )
var2 = np.matrix( [ np.random.randint(0,20,3) , np.random.randint(20,40,3), np.random.randint(20,40,3) , np.random.randint(20,40,3) , np.random.randint(20,40,3) ] )


print( var1 )
print()
print( var2 )
print()

print( var1* var2 )   # matrix condition is that columns in matrix A should be equal to row in matrix B for A X B matrix
print()
print( var1.dot(var2) )  # final matrix will be ( rows in matrix A , column in matrix B )
```

```python
# lets learn about transpose function

var3 = np.matrix( np.random.randint(0,20, (3,5)) )
print( var3 )
print()

a = np.transpose( var3 )  # it will return the transpose of var3
print( a )
print()

b = var3.T # works as same to transpose the array
print( b )
```

```python
# now lets learn about inverse 
import time

start = time.perf_counter()

var4 = np.matrix( np.random.randint(0,20, ( 5,5 )) )
print( var4 )
print()

a = np.linalg.inv( var4 )   # it will return the inverse of var4
print( a )
print()

end = time.perf_counter()
print( f' {(end - start):.6f} ' )
```

```python
# lets learn about power
var4 = np.matrix( np.random.randint(0,20, ( 5,5 )) )
a = np.linalg.matrix_power( var4 , 3 )  # ( array , power )   , n = 0 means identity , n < 0 represents inverse

print( a )
```

```python
# lets learn about determinant

var5 = np.matrix( np.random.randint(0 , 20 , ( 4 , 4 ) ) )
print( var5 )
a = np.linalg.det( var5 )
print(f'{a:.3f}')
```
