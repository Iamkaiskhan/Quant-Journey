```python
# lets learn about shuffle function

import numpy as np

arr = np.array( np.random.randint( 0,20,10 ) )
print( arr )
print()

np.random.shuffle( arr )   # it shuffles the original array

print( arr )
```

```python
# to filter the unique elements from array
arr1 = np.array( np.random.randint( 0,20,10 ) )
print( arr1 )
print()

x = np.unique( arr1 )   # it gives unique elements and elements are in increasing order
print( x )
print()

y = np.unique( arr1 , return_index = True , return_counts = True ) # return_index will return the index of the elements and the return_count will return the number of repeated elements
print( y )
```

```python
# lets learn about resize function

arr2 = np.array( np.random.randint(0,20,10) )
print( arr2 )
print() 

y = np.resize( arr2 , ( 5,2,1 ) )
print( y )
```

```python
# lets learn about insert fucntion 
arr3 = np.array( np.random.randint(0,10,5) )
print( arr3 )
print()

v = np.insert( arr3 , 2 , 30 )   # insert( arr3 , index , element ) , insert fucntion only takes integer values
print( v )

```

```python
# insert() function for 2 d 

arr4 = np.array( [ np.random.randint(0,10,6) , np.random.randint(10,20,6) ] )
print( arr4 )
print()

q_column = np.insert( arr4 , 1 , 50 ,  axis = 0 ) # in this case in each column with index 1 will have value as 50 of element
print( q_column )
print()

q_rows = np.insert ( arr4 , 1 , 22 , axis = 1 ) # in this case in each row with index 1 will have the value as 22 of element
print( q_rows )
```
