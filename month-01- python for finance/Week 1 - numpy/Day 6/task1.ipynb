```python
import numpy as np

arr = np.array ( np.random.randint(1,15,10) )
print( arr )
print()
# lets learn about where() function which will return the index of the elements if the condition is met
x = np.where( arr%2 == 0 )
print ( x )
```

```python
# to learn how to sort an array

arr1 = np.array( np.random.randint(1,15,10) )
print( arr1 )
print()
arr1_sort = np.sort( arr1 )
print( arr1_sort )


# lets learn to place element in the array if the array is already sorted
# searchsorted() return the index

x = np.searchsorted( arr1_sort , [ 2,34 ,5,8 ] )  # we can pass the elements that we want to append as list also if we want to append more than 1 element
print()
print(x)
print()
print( arr1_sort )
```

```python
# to sort 2 d array
arr2 = np.array( [ np.random.randint(1,15,10) , np.random.randint(1,15,10) , np.random.randint(1,15,10) ] )
print( arr2 )
print()

arr2_sort = np.sort( arr2 )    # it will sort the array in ascending order for each row
print( arr2_sort )
```
