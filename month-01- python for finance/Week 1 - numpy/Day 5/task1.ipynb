```python
# lets learn about join array

import numpy as np

arr1 = np.array( np.random.randint(0,100,8) )
arr2 = np.array( np.random.randint(0,100,8) )

print(arr1)
print()
print(arr2)
print()

arr_merge1 = np.concatenate((arr1,arr2))
print(arr_merge1)
```

```python
# lets learn join in 2d 
arr3 = np.array( [ np.random.randint(0,100,4) ,  np.random.randint(0,100,4) ,  np.random.randint(0,100,4) ] )
arr4 = np.array( [  np.random.randint(0,100,4) ,  np.random.randint(0,100,4) ,  np.random.randint(0,100,4) ])

print(arr3)
print()
print(arr4)
print()

arr_new_column = np.concatenate( (arr3,arr4) , axis = 0 )   # we have to define axis for this
arr_new_row    = np.concatenate( (arr3,arr4) , axis = 1 )

print( 'The array along column is\n' , arr_new_column ) 
print()
print( 'The array along row is\n' , arr_new_row ) 
```

This below is the example of join in 2 d


```python
# lets learn about stack() function which will help to convert 1 d to 2 d array
arr4 = np.array ( np.random.randint(0,100,7) )
arr5 = np.array ( np.random.randint(0,100,7) )

arr_stack_column = np.stack( (arr4,arr5) , axis = 0 ) # it will merge across columns , we can also write np.vstack( (arr4,arr5) ) which will behave as for column
arr_stack_row = np.stack( (arr4,arr5) , axis = 1 )   # it will merge across row,     we can also write np.hstack( (arr4,arr5) ) which will behave as for row

print( arr4 )
print()
print( arr5 )
print()

print( 'The stack across column is\n', arr_stack_column )
print()
print('The stack across  row is\n' , arr_stack_row )
print()
```

```python
# lets learn about splitting in a array

arr6 = np.array( np.random.randint( 0,100,13 ) )
print( arr6 )
print()

arr_split = np.array_split( arr6 , 5 )    # here let say i am saying to have partition with 5 elements each and total element in arr6 is 13 , so divide 13 by 5 so remainder is 3 and quotient is 2 it means , 2 elements in each array but remainder is 3 which means extra 3 elements are there so first 3 arrays will have 1 extra element of remainder
# the data type of array_spilt() is "LIST"

print( arr_split )
print( arr_split[2] )
print( type(arr_split[2]) ) # the arr_spilt is list but inside that elements of arrays are of numpy array data type
```

```python
# lets learn about split in 2 d

arr7 = np.array( [ np.random.randint( 0,100,5 ) , np.random.randint( 0,100,5 ) , np.random.randint( 0,100,5 ) ] )
print( arr7 )
print() 

arr_2d_column = np.array_split( arr7 , 5 , axis = 1 )
arr_2d_row = np.array_split( arr7 , 5 , axis = 0 )


print('The split across columns')
print( arr_2d_column )
print()
print('The split across rows')
print( arr_2d_row )

arr_2d = np.array_split( arr7 , 4 ) # it will bydefault spilt across rows
print()
print(arr_2d)
```
