# Creating numpy arrays



```python
# we can create array in following ways
import numpy as np

# Method 1
x= [1,2,3,4]  # we created list
y= np.array(x)
print(y)

# Method 2 
z = np.array([1,2,3,4,5])
print(z)
z       # this will print tyhe type of array along with the array    O/P: array([1, 2, 3, 4, 5])
print(type(z))

```

```python
# array input 
import numpy as np
l=[]
print('Enter the input:')
for i in range(5):
    num=input()      # the input is taken as string so it will stored as string , not integer
    l.append(num)


print(np.array(l))   
```

```python
# we will learn to find the dimension of array
x=[1,2,3,4,5,6,7]
y = np.array(x)
print(y)
print(y.ndim)   # it will tell us the dimension
```

```python
# 2-D array
x = np.array([[1,2,3,4,5],[5,6,3,56,5],[5,6,3,56,5]])  # we can pass as many as arrays in any dimension
print(x)
print(x.ndim)


# 3-D array
y = np.array([[[3,2,1],[3,4,5]]])
print(y)
print(y.ndim)
```

```python
# to create array of n dimension
n=int(input("Enter the dimension:"))
x=np.array([4,5,4,6],ndmin=n)   # ndmin is used to classify the dimension
print(x)
print(x.ndim)
```

```python
# learning about zeroes ones empty

x=np.zeros(10)   # it will create array with 0 in each elements, here number of elements are 10
print(x)

y = np.ones(5)   # it will create array with 1 in each elements, here number of elements are 5
print(y)

z = np.empty(15) # it will create array with each element being empty , here number of elements are 15 
print(z)          # it will store junk values as of now in each element

arr = np.array([2,3,56.435])
print(arr.dtype)  # it will print the data tpye
a = arr.astype(np.int32)  # hence we are converting it to int32 so the 50.435 will convert to int i.e 56  ## we use astype() for casting
print(a)
print(a.dtype)
```
