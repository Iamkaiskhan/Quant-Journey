```python
import numpy as np

x=np.array([1,2,3,4])    # it is array
print(x)

y=[1,2,3,4]     # it is list
print(y)

print('The type of x is {} and y is {}'.format(type(x),type(y)))

```

# Difference between numpy array and list in python

1) Data Type storage:
In numpy we can either work on only string or only in numbers but in list we can work on both integers and string combine

2) Importing module:
In list we dont have to impoort module to work on it but in numpy we have to install the package

3) Numerical Operation:
In list the numerical operation is very less efficient as compare to numpy
We cannot give matrice format to list 

4) Memory consumption:
Numpy consumes less memory as compare to list

5) Speed:
Numpy works faster as compare to list
