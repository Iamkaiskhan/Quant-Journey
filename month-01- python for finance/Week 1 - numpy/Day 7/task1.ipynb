```python
import numpy as np

rows = int(input('Enter the number of rows:'))
columns = int(input('Enter the number of columns:'))

data = np.array( np.random.randint( 0,100, (rows,columns)))

filename = 'numbers.txt'

np.savetxt( filename , data , delimiter = ',' , fmt= '%d')
```

```python
a = np.loadtxt( 'numbers.txt' , delimiter = ',')
b = a.astype('int32')
print(b)
print()
```

```python
# now lets use boolean

boolean = b > 50
print( boolean )
print()
for row in boolean :
    for column in row :
        if column == True:
            print(column)
```
