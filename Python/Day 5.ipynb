```python
# lets learn to use number system
print(0b110110010)
print(0b1101011)  # prints 107   BINARY SYSTEM
print(0xFB + 0b10)  # prints 253   HEXADECIMAL SYSTEM
print(0o15)  # prints 13    OCTAL SYSTEM
```

```python
# Lets learn about random function
import random

print(random.randrange(10,30))

list=['a','g','k','d','z']
print(random.choice(list)) # get random item from list1

random.shuffle(list)  
print(list)  # Print the shuffled list1

print(random.random())
```

```python
import random
random.seed(10)  # if we give any seed value ,depending on the value of seed, the random varaibles will be same ,everytime we run the function

print(random.random())
print(random.random())
print(random.random())
```

```python
# lets learn about python math functions
import math
print(math.pi)   # prints the pi value 
print(math.cos(math.pi))  # cos(x)
print(math.log10(100))   # logarithmic function
print(math.exp(2))
print(math.factorial(10))
print(math.sinh(99))
```
