```python
#lets learn about tuple 
numbers=(1,9,-4,'k')
print(numbers)

# we can also create tuple in the below manner
x= tuple(('Kais','Quant','Trading'))
print(x)

#we can also create a empty tuple 
empty=()
print(f'the type of {empty} is {type(empty)}')
```

```python
# Indexing in tuples 
language=('python','c++','java')

print(language[0])   # the 1st element
print(language[2])   # the 3rd element


# random.shuffle(language)       IF U WILL RUN THIS CODE , IT WILL SHOW ERROR, WHICH MEANS TUPLE IS IMMUTABLE
# print(language)

language[0]='Kais'   # will show error ,tuple is immutable
```

```python
# iterate in tuple
numbers=(1,9,28,0,7,9,2,0,9,1)
for x in numbers:
    if x==9: continue
    print(x)
    
```

```python
# Check if an Item Exists in the Tuple
colors = ('red', 'orange', 'blue')
print( 'orange' in colors)
print( 'pink' in colors)


# to delete the tuple
del colors

```

```python
# important note
test=('Hello')
print(type(test)) # it will print type as STRING not TUPLE 

# the 1st method:
test= tuple(('Hello'))
print(type(test))

# the 2nd method
test=('Hello',)
print(type(test))
```
