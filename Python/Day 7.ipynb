```python
#methods of python strings

# 1. to convert the string to upper case
name='KaIs'
name_upper=name.upper()
print(name_upper)

# 2. to convert to lower cases
name_lower=name.lower()
print(name_lower)

# 3. to know the index of the character
sentence='My name is Kais Khan'
index=sentence.index('Kais Khan')
#if u ever confuse about lower upper case to find index
indexx=(sentence.upper()).index('KAIS KHAN') # we can convert the whole string to upper or lower cases then iterate to find index

print(index)
print(indexx)
```

```python
# now lets learn about sets 
# A set cannot have same elements
test={'Kais',20,'Mumbai',2005,'Male'}
print(test)

# to create a empty set
empty_set=set()
print(empty_set)
print(type(empty_set))

empty_dictionary={}
print(type(empty_dictionary))


set1={1,2,9,2,9,1,0,9,28,9,10,23}
print(set1)   # the duplicate are removed ,hence the set cannot store the duplicate values
```

```python
# sets are mutable but are unordered so they cannot be changed using indexing or slicing

numbers={ 1,60,5,6,18,4}
print(f'The inital set is {numbers}')
# now to add element in set
numbers.add(30)
numbers.add('Kais')
print(f'The updated set is {numbers}')



# we can update the set which means that we can add the items of list,tuple,set,etc to a set 
chem_branch=['Fluid Mechanics','Thermodynamics']
mining_branch={'Mine Ventilation','Rock Mechanics'}

mining_branch.update(chem_branch)

print(mining_branch)
```

```python
#we can also discard any element from the set
language={'Java','Python','C'}

language.discard('Java')
print(language)


#lets learn about functions inside set 
dummy={10,'A',54,70,5,'P','Kais','Quant',5,6,40,9}

numbers={10,9,75,9,-45,960,52,11}
# to find the min of the set  (so the set must be completely numeric)
minimum=min(numbers)
print(minimum)

# to find the max of the set (so the set must be completelt numeric)
maximum=max(numbers)
print(maximum)


# to find the sum of the set 
summation= sum(numbers)
print(summation)

# to sort the set in ascending order ( if we have to create a another set for this task)
sort=sorted(numbers)
print(sort)

# to sort in descending order 
sort=sorted(numbers, reverse=True)
print(sort)

# iterate in set 
for x in numbers:
    print(x)
```

```python
# first set 
A={1,3,5,7}

# second set
B={0,2,4,6}

# Method 1 to take union of set
C= A|B
print(A,'\n',B,'\n',C,sep='')

# Method 2 to take union of set
c= A.union(B)
print(c)
```

```python
set1={1,23,5,79,1,6,0,58,6,4,6}
set2={0,9,1,6,7,5,6,9,0,6,1,6}

# method 1 to take intersection
set3= set1 & set2
print(set3)

# method 2 to take intersection
set3= set1.intersection(set2)
print(set3)

## now we will learn to find the difference which means the elements in one set which the other dosent include
# method 1
diff= set1 - set2 
print(diff)

# method 2
diff= set1.difference(set2)
print(diff)


## now to find the symmetric difference which means that the elements excluding the intersection(common elements), it means (A union B)-(A intersection B)


# method 1
symmetric= set1 ^ set2
print(symmetric)

# method 2
symmetric= set1.symmetric_difference(set2)
print(symmetric)
```
