```python
my_list = ['p', 'r', 'o', 'g', 'r', 'a', 'm']
print(my_list)

# get a list with items from index 2 to index 4 (index 5 is not included)
print('my_list[2:5]=',my_list[2:5],sep='')

# get a list with items from index 2 to index -3 (index -2 is not included)
print('my_list[2:-3]=',my_list[2:-2])

print(my_list[:])

# get a list with items from index 0 to index 2 (index 3 is not included)
print('my_list[0:2]=',my_list[0:2])

import random
random.shuffle(my_list)
print(my_list)
```

```python
#Omitting Start and End Indices in Slicing
my_list = ['p', 'r', 'o', 'g', 'r', 'a', 'm']
print(my_list)

# get a list with items from index 5 to last
print('my_list[5:]=',my_list[5:])

# get a list from the first item to index -5
print('my_list[:-4]=',my_list[:-4])

# get a list from start to end items
print('my_list[:]=',my_list[:])
```

```python
fruits = ['apple', 'banana', 'orange']
print(f'The original list is: {fruits}')
#Add Elements to a Python List
fruits.append('mango')

print(f'the updated list is: {fruits}')

#if we wish to append the item in the particular index 
fruits.insert(2,'papaya')    # (index,item name)
print(f'the updated list is: {fruits}')
```

```python
#Add Elements to a List From Other Iterables
numbers_1=[1,3,5]
print(f'{numbers_1}')

numbers_2=[2,4,6]
print(f'{numbers_2}')

numbers_1.extend(numbers_2)
print(f'{numbers_1}')
```

```python
# Change List Items
colors=['Red','Yellow','Black']
print(colors)

# change the first item to 'Purple'
colors[0]='Purple'

# change the third item to 'Blue'
colors[2]='Blue'


print(f'the updated list is {colors}')
```

```python
# we will learn to remove item from list
numbers=[1,2,3,8,9,7,0,8]

numbers.remove(8)    # it will remove the first ever available item from the list
print(numbers)
index=numbers.index(3)
print(index)

#if we want to delete more than one item
names = ['John', 'Eva', 'Laura', 'Nick', 'Jack']
del names[1]
print(names)

# delete items from index 1 to index 2
del names[1:3]
print(names)

# To know the number of elements
print(len(names))

# delete the entire list
del names
```
