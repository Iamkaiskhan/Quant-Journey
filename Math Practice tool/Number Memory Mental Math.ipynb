```python
import numpy as np
from math import pow
import time

# no subtraction yet
number_of_digit = 1
iteration_cycle = 3
sleep_time = 1
a = 0
sum = 0
status = True
while status :
    x = np.array( np.random.randint( pow( 10, number_of_digit - 1 ) , pow( 10, number_of_digit )-1 , 1 ))
    sum += x
    time.sleep( sleep_time )
    print( x ,end= '')
    print()
    a+=1
    if a == iteration_cycle :
        time.sleep( sleep_time )
        print("The sum is:",sum )
        a = 0 
        sum = 0
        status = False
```
