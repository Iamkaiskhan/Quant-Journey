```python
# division practising tool 
import numpy as np
number_of_digit = 3
denominator_number = 2

class Invalid(Exception) :
    pass

while True :
    x = np.array ( np.random.randint( 1 , 10 ** number_of_digit , 1 ) )

    if x % denominator_number == 0 :
        try :
            print(f'{x}/{denominator_number}')
            
            ans = int(input())
            
            if ans != (x / denominator_number) :
                raise Invalid


        except Invalid :
            print('Answer is:', x / denominator_number )

    else :
        continue
```
