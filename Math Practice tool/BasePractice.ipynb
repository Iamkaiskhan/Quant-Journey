```python
import numpy as np
import math
import time



a = 0
total_time = 0

status = True 
while status :
    base = np.array ( np.random.randint( 1 , 5 , 1 ) )

    var = np.array( np.random.randint( math.pow( 10 , base ) - 10 , math.pow( 10 , base ) , 2 ) )
    print( var )
    
    start = time.perf_counter()
    answer = int(input())
    stop = time.perf_counter()

    time_taken = stop - start
    if answer == var[0] * var[1]:
        print('correct')
        a+=1
        total_time += time_taken

    else :
        print(f'Question solved :{a}')
        print(f'Total time taken : { total_time } ' )
        print(f'Average time : { total_time / a } ' )

        status = False
```
