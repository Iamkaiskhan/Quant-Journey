```python
import numpy as np
import time

status = True
solved = 0
total_time = 0
while status :
    multiple = np.random.randint( 2 , 13 , 1 )
    arr = np.array( np.random.randint( 10 , 10000 , 1 ) )
    start = time.perf_counter()
    print(f' { arr } * { multiple } ')

    answer = arr * multiple 
    given_answer = int(input())
    stop = time.perf_counter()
    time_taken = stop - start
    if answer == given_answer :
        print(f'Time taken : { ( time_taken ):.2f}')
        print('Correct')
        solved+=1
        total_time += time_taken
        continue

    else :
        print('incorrect')
        print('Solved questions',solved)
        print('Total time', total_time)
        print('avg time per question', total_time/solved)
        status = False

```
