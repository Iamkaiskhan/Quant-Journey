```python
# Why does 0.1 + 0.2 == 0.3 evaluate to False? How do you compare floats safely?
# Floating-point representation error. 0.1 + 0.2 actually gives 0.30000000000000004
# so the == check fails. Never compare floats with ==. Use a tolerance:
import math
k=math.isclose(0.1+0.2,0.3)
print(round(7.4))
print(k)


# slicing : [start : stop : step]
# if step is negative the start is from last index(if the start is not mentioned)
# if step is positive the start is from beginning index which is 0(if the start index is not mentioned)
prices = [100, 101, 102, 103, 104] 
print(prices[ ::1])    
print(prices[ ::-1])   #reverse 
print(prices[ :: -2])  #reverse
```

```python

def sum_numbers(numbers):
    return sum(numbers)

a=int(input('How many numbers:\n'))
numbers=[]
for x in range(a):
    number=int(input(f'enter the {x+1} number:'))
    numbers.append(number)

result=sum_numbers(numbers)
print(result)
```

```python
# function to sum any number of argument
def add_all(*numbers):
    return sum(numbers)

# pass any number of arguments
print(add_all(1, 2, 3, 4))   
```

```python
# ** treats the parameter as dictionary
def greet(**words):
    for key,value in words.items():
        print(f' {key} : {value} ')

greet( name='Kais', branch='Mining')
```

```python
# function call with two arguments
def add_numbers(a=7,b=3):
    return a+b

# function call with two arguments
print(add_numbers(2,4))

# function call with one arguments
print(add_numbers(b=8))      # a parameter remains same, only b takes the value which we gave

# function with no call arguments
print(add_numbers())       # takes the default values of a and b
```

```python
# In keyword arguments, arguments are assigned based on the name of the arguments. For example,

def name(first_name,last_name):
    print('First name',first_name)
    print('Last name',last_name)

name(last_name='Khan',first_name='Kais')   # see clearly the names are arranged in output as per parameter defined in function
```

```python
# program to find sum of multiple numbers 
def summation(*numbers):
    result=0
    for num in numbers:
        result+=num

    return result

answer=summation(4,34,5,35,6,3)
print(answer)


a=(45,54,643,6,734,7)
sum=0
for num in a:
    sum+=num

print(sum)
print(type(a))


# to add element in tuple | IMPORTANT
a=(3,54,7,34,5,0)
print(a)

# method 1
a=a+ (7,)
print(a)

# method 2
a= a + tuple((399,))
print(a)
```
