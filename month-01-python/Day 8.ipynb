```python
# lets learn about dictionary

# creating a dictionary 

country_capital={
    'Germany': 'Berlin',
    'Canada' : 'Ottawa',
    'England': 'London'
}

print(country_capital)

# KEYS of dictionary are immutable
```

```python
# Create Dictionary Using keyword arguments only

numbers = dict(x=5,y=10)
print(f'the numbers are {numbers}')

empty=dict()
print(empty)

# the dictionary are immutable
# 1. the valid dictionary, INTEGER AS KEY
dict1={
    1: 'Kais',5:'Khan', 'name':'Me'
}
print(dict1)

# 2. we can also use TUPLE as KEY
dict2= {
    (1,2):'One Two',3:'Three'
}
print(dict2)

# 3. we cannot use LISTS as KEY , but can be use as a VALUE
"""
dict3= {
    (1,2):'Kais Khan',[5,9,1,5]:'Hey'
} 
print(dict3)
"""

# 4. use KEY as STRING or NUMERIC and list as VALUE
dict4={
    'list':[80,48,'Hey']
}
print(dict4)
```

```python
# if there are duplicate key with different values then the last key with the value will be considered
hogwarts_houses = {
    "Harry Potter": "Gryffindor",
    "Hermione Granger": "Gryffindor",
    "Ron Weasley": "Gryffindor",
    # duplicate key with a different house
    "Harry Potter": "Slytherin"
}

print(hogwarts_houses)
```

```python
# we can use two methods to access the dictionary items

# method 1
country_capitals = {
  "Germany": "Berlin", 
  "Canada": "Ottawa", 
  "England": "London"
}

print(country_capitals['Germany'])


# method 2 is that we can use the get() function for this task , get(key,value)
print( country_capitals.get('Germany'))
print( country_capitals.get('Germany',21))   

concept= '''
So the get function work in a manner like if the KEY is found then it will not cross verify our given VALUE which here i gave 21 
and if the KEY is not in the dictionary then it will return the value of VALUE and if we didnt gave the value of VALUE then it 
will return None
'''


profile={
    'name':'Kais','age':20,'Branch':'Mining'
}
print( f'the name is {profile.get('name')}')
print( f'the age is {profile.get('age')}')
print( f'the branch is {profile.get('Branch')}')
```

```python
# to add item in the dictionary
country_capitals = {
  "Germany": "Berlin", 
  "Canada": "Ottawa", 
  "England": "London"
}

country_capitals['India']='Delhi'
print(country_capitals)

# now to delete item from the dictionary
# we can use the del function
del country_capitals["Germany"]
print(country_capitals)
# now lets add it again 
country_capitals['Germany']='Berlin'

# we can also use pop()
remove=country_capitals.pop('Canada','Kais')   # first it will look at the key value,if the key value is not found then it will look at the VALUE
print(remove)
print(country_capitals)

# to change the dictionary item 

country_capitals['England']='LONDON'
print(country_capitals)
```

```python
# we can also use the update() function to add or update in a dictionary
marks={
    'maths':80,'chemistry':67
}

marks.update({'Physics':79})
print(marks)


# OR we can also do the same by creating another dictionary
a={ 'Physics':99}
marks.update(a)

print(marks)
```

```python
# iterate in dictionary
country_capitals = {
  "Germany": "Berlin", 
  "Canada": "Ottawa", 
  "England": "London"
}

# to print the name of the country
for country in country_capitals:
    print(country)


# to print the capital of the country
for country in country_capitals:
    capital=country_capitals[country]
    print(f'The name of Country and its Capital is {country}\{capital}') 
    
```

```python
# to find the length of the dictionary 
country_capitals = {"England": "London", "Italy": "Rome"}

# get dictionary's length
print(len(country_capitals))   # Output: 2

numbers = {10: "ten", 20: "twenty", 30: "thirty"}

# get dictionary's length
print(len(numbers))            # Output: 3

countries = {}

# get dictionary's length
print(len(countries))          # Output: 0
```

```python
file_types = {
    ".txt": "Text File",
    ".pdf": "PDF Document",
    ".jpg": "JPEG Image",
}

# use of in and not in operators
print(".pdf" in file_types)       # Output: True
print(".mp3" in file_types)       # Output: False
print(".mp3" not in file_types)   # Output: True
```
