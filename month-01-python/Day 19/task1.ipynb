```python
# we will learn about date time
import datetime as dt
current_date=dt.date.today()
print(current_date)

# to store date

date1=dt.date(2026,9,20)          # (year,month,day)
print(date1)

# to print year month and day separately
print('Year',date1.year)
print('Month',date1.month)
print('Day',date1.day)

```

```python
# now lets learn about time module

import datetime as dt

time1=dt.time(12,56,13,2345)  # (hour,minutes,seconds,microseconds)  
print(time1)

# to print hours minutes seconds separately
print('Hour',time1.hour)
print('Minutes',time1.minute)
print('Seconds',time1.second)
print('Microseconds',time1.microsecond)
```

```python
# now to work with both date and time combine
import datetime as dt 

datetime_obj=dt.datetime(2026,9,20,20,30,17)   # (year,month,day,hour,minute,seconds,microseconds)

print(datetime_obj)

# to print date time separately
print(datetime_obj.date())

print(datetime_obj.time())



# to print the current datetime

current_datetime=dt.datetime.now()

print(current_datetime)



# we can also find the delta which means time difference between 2 periods

time2=dt.datetime.now()
time3=dt.datetime(2025,9,20)

difference= time2 - time3
print(difference)


# we can use strf function for customizing the date
current_datetime=dt.datetime.now()
strftime=current_datetime.strftime('%A,%B %d,%Y')

print(strftime)
```
