```python
# we will solve example to print fibonacci using recursion

def fibonacci(n):
    if n<=1:
        return n

    else:
        return fibonacci(n-1)+fibonacci(n-2)



while True:
    user_input=input('Enter the number of terms:')

    try:
        num=int(user_input)
    except:
        print('That\'s not a valid input')
        continue

    if num<=0:
        print('That\'s not a valid input')
        continue

    break

    
series=[]

print('The fibonacci series is:')
for i in range(num):
    series.append(fibonacci(i))

print(series)


    
```
